```python
#Requirements 1: 
mapping_dict = dict(zip(customer_master_df["Old_Cust_Code"], customer_master_df["New_Cust_Code"]))
wip_ageing_df["New_Cust"] = wip_ageing_df["Cust"].map(mapping_dict).fillna(wip_ageing_df["Cust"])
wip_ageing_df.head()
```

```python
# the second requirements make me confused that I view the data and I realize the value in the "New_Cust" have 6 -digits before the "-" character. Therefore, I
# to get 6 digits.
# using regex to extract alphabets character and numbers
alphabet_character_regex = '([A-Za-z]+)' #if you want to get alphabet characters
nnumeric_character_regex = '(\d+)' #if you want to get numeric characters
wip_ageing_df_step2 = wip_ageing_df.copy()
wip_ageing_df_step2['6-digit'] = wip_ageing_df_step2["New_Cust"].str.split("-").str[0].str.extract(nnumeric_character_regex)
wip_ageing_df_step2.head()
```

```python
#the third requirement
# Sort and drop duplicates base on ManuPartNo
wip_ageing_df_step2 = wip_ageing_df_step2.sort_values(by=['ManuPartNo', 'DateIn'], ascending=[True, False])
wip_ageing_df_step3 = wip_ageing_df_step2.drop_duplicates(subset=["ManuPartNo"], keep="first")
wip_ageing_df_step3.head()
```

```python
wip_ageing_df_step4 = wip_ageing_df_step3.copy()
# this step I will calculate the timestamp today and add  a new column ts_time of DateIn. I will Subtract both and get the result 
# and transform to months ttype and classtification into age buckets.  
AVERAGE_DAY_PER_MONTH = 30
wip_ageing_df_step4['ts_time'] = pd.to_datetime(wip_ageing_df_step4['DateIn']).dt.normalize()
today_ts = pd.Timestamp.today().normalize()
wip_ageing_df_step4['months_diff'] = ((today_ts - wip_ageing_df_step4['ts_time']).dt.days / AVERAGE_DAY_PER_MONTH).round(2)  # Average days in a month
# Define the Swith function to categorize into age buckets
bins = [0, 9, 12, 13, 24, 25, float('inf')]  # Khoảng: <9, 9-12, 12-13, 13-24, 24-25, >25
labels_value = ["BELOW09MTHS", ">09TO12MTHS", ">12TO13MTHS", ">13TO24MTHS", ">24TO25MTHS", ">25MTHS"]
wip_ageing_df_step4['Age_Bucket'] = pd.cut(wip_ageing_df_step4['months_diff'], bins=bins, labels=labels_value, right=False)  # right=False để <9 là exclusive

wip_ageing_df_step4.head()


```

```python
# Data Quality Check Functions 
# I will check some case to evaluate the data quality of the final dataframe. 
# #I will check some case such as null value, duplicate value, 6-digit value and numeric only value.
import pandas as pd
import numpy as np

def check_not_null(df, columns): 
    """
    Check for null values in specified columns.
    Returns a dict with column names as keys and count of nulls as values.
    """
    null_counts = {}
    for col in columns:
        if col in df.columns:
            null_counts[col] = df[col].isnull().sum()
        else:
            null_counts[col] = f"Column '{col}' not found in DataFrame"
    return null_counts

def check_duplicate_rows(df): #all columns
    """
    Check for completely duplicate rows.
    Returns a dict with total duplicates and list of duplicate indices.
    """
    duplicates = df[df.duplicated(keep=False)]
    duplicate_indices = duplicates.index.tolist()
    return {
        'total_duplicates': len(duplicate_indices),
        'duplicate_indices': duplicate_indices,
        'duplicate_rows': duplicates
    }

def check_exactly_6_digits(df, column): # To check 6-digit	
    """
    Check if values in the column are exactly 6 digits (numeric characters).
    Returns a dict with count of valid/invalid and list of invalid indices.
    """
    if column not in df.columns:
        return f"Column '{column}' not found in DataFrame"
    
    # Convert to string and check length and if all digits
    is_6_digits = df[column].astype(str).str.match(r'^\d{6}$')
    invalid_indices = df[~is_6_digits].index.tolist()
    return {
        'valid_count': is_6_digits.sum(),
        'invalid_count': (~is_6_digits).sum(),
        'invalid_indices': invalid_indices,
        'invalid_values': df.loc[invalid_indices, column].tolist()
    }

def check_numeric_only(df, column): # To check ManuPartNo, SvcOrder, Tot_Amt
    """
    Check if values in the column contain only numeric characters (digits).
    Returns a dict with count of valid/invalid and list of invalid indices.
    """
    if column not in df.columns:
        return f"Column '{column}' not found in DataFrame"
    
    # Check if string representation is all digits
    is_numeric = df[column].astype(str).str.isdigit()
    invalid_indices = df[~is_numeric].index.tolist()
    return {
        'valid_count': is_numeric.sum(),
        'invalid_count': (~is_numeric).sum(),
        'invalid_indices': invalid_indices,
        'invalid_values': df.loc[invalid_indices, column].tolist()
    }

def run_data_quality_checks(df, columns_to_check_null=None, digit_column=None, numeric_column=None):
    """
    Run all data quality checks and return a summary report.
    """
    report = {}
    
    # Check not null
    if columns_to_check_null:
        report['null_checks'] = check_not_null(df, columns_to_check_null)
    

    # Check duplicates
    report['duplicate_rows'] = check_duplicate_rows(df)

    # Check 6 digits
    if digit_column:
        report['6_digits_check'] = check_exactly_6_digits(df, digit_column)

    # Check numeric only
    if numeric_column:
        report['numeric_only_check'] = check_numeric_only(df, numeric_column)

    return report


columns_to_check = ['New_Cust', '6-digit', 'DateIn', 'Age_Bucket']
quality_report = run_data_quality_checks(
    wip_ageing_df_step4, 
    columns_to_check_null=columns_to_check, 
    digit_column='6-digit', 
    numeric_column='6-digit'
)
print(quality_report)
```
